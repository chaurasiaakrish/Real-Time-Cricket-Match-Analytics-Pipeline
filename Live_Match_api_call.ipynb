```python
import pandas as pd
import json

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
```

```python
import requests

url = "https://api.cricapi.com/v1/match_scorecard"

params = {
    "apikey": "XXXXXXX",
    "id": "XXXXXXXXX"
}

response = requests.get(url, params=params)
```

```python
data=response.json()
with open("match_snapshot.json", "w") as f:
    json.dump(data, f)
    print(data)
```

```python
def extract_first_batting(data):
    batting = data["data"]["scorecard"][0]["batting"]
    
    rows = []
    
    for player in batting:
        rows.append({
            "Batsman": player["batsman"]["name"],
            "Runs": player["r"],
            "Balls": player["b"],
            "4s": player["4s"],
            "6s": player["6s"],
            "StrikeRate": player["sr"],
            "Dismissal": player.get("dismissal-text", "NA")
        })
    
    return pd.DataFrame(rows)

ind_batting_df = extract_first_batting(data)

print(ind_batting_df.to_string())
```

```python
def extract_first_bowling(data):
    bowling = data["data"]["scorecard"][0]["bowling"]
    
    rows = []
    
    for bowler in bowling:   # change here
        rows.append({
            "Bowler": bowler["bowler"]["name"],
            "Overs": bowler["o"],
            "Maiden": bowler["m"],
            "Runs": bowler["r"],
            "Wickets": bowler["w"],
            "No-Ball": bowler["nb"],
            "Wide": bowler["wd"],
            "Economy": bowler["eco"]
        })
    
    return pd.DataFrame(rows)

pak_bowling_df = extract_first_bowling(data)
print(pak_bowling_df.to_string())
```
